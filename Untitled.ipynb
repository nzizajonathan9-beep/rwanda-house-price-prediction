## Model Evaluation

```python
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

print("Predictions generated successfully!")
```

```python
def calculate_metrics(y_actual, y_pred, number_of_predictors):
    
    r2 = r2_score(y_actual, y_pred)
    
    adjusted_r2 = (
        1 - (1 - r2) *
        (len(y_actual) - 1) /
        (len(y_actual) - number_of_predictors - 1)
    )
    
    mae = mean_absolute_error(y_actual, y_pred)
    
    rmse = np.sqrt(
        mean_squared_error(y_actual, y_pred)
    )
    
    return r2, adjusted_r2, mae, rmse
```

```python
train_r2, train_adj_r2, train_mae, train_rmse = calculate_metrics(
    y_train,
    y_train_pred,
    len(feature_names)
)

print("Training R²:", train_r2)
print("Training Adjusted R²:", train_adj_r2)
print("Training MAE:", train_mae)
print("Training RMSE:", train_rmse)
```

# Calculate testing metrics

```python
test_r2, test_adj_r2, test_mae, test_rmse = calculate_metrics(
    y_test,
    y_test_pred,
    len(feature_names)
)

print("Testing R²:", test_r2)
print("Testing Adjusted R²:", test_adj_r2)
print("Testing MAE:", test_mae)
print("Testing RMSE:", test_rmse)
```

```python
evaluation_table = pd.DataFrame({
    "Dataset": ["Training", "Testing"],
    "R²": [train_r2, test_r2],
    "Adjusted R²": [train_adj_r2, test_adj_r2],
    "MAE (Million RWF)": [train_mae, test_mae],
    "RMSE (Million RWF)": [train_rmse, test_rmse]
})

evaluation_table
```
